```python
from pathlib import Path

folder_name = "enveda-CASMI26-molecule-id-mass-spectra"

candidates = [
    Path.cwd() / folder_name,
    Path.cwd().parent / folder_name,
]

data_dir = next((p for p in candidates if p.exists()), None)

if data_dir:
    print("Found data folder:", data_dir)
    print("Contents:", [f.name for f in data_dir.iterdir()])
else:
    print("Data folder not found. We'll check its location.")
```

```python
import pandas as pd

train_path = data_dir / "spectra_train" / "part-000.parquet"

df = pd.read_parquet(train_path)

print("Data loaded successfully!")
print("Rows and columns:", df.shape)

df.head()
```

```python
import matplotlib.pyplot as plt
import numpy as np

counts = df["inchikey14"].value_counts()

print("Total spectra:", len(df))
print("Unique molecules:", df["inchikey14"].nunique())
print("Molecules with 2+ spectra:", (counts >= 2).sum())
print("Molecules with 5+ spectra:", (counts >= 5).sum())
print("Molecules with 10+ spectra:", (counts >= 10).sum())

print("\nClass frequency distribution:")
print(counts.describe(percentiles=[.25, .5, .75, .9, .95, .99]))

# Plot distribution of spectra per molecule
plt.figure(figsize=(9, 5))
plt.hist(counts.clip(upper=30), bins=30)
plt.xlabel("Number of spectra per molecule (30 = 30+)")
plt.ylabel("Number of molecules")
plt.title("Molecular Class Distribution")
plt.tight_layout()
plt.show()

# Inspect the 20 most represented molecules
print("\nTop 20 molecular classes:")
print(counts.head(20))
```
